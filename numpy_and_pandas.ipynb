loc and iloc



iloc default index base

loc manual index base





```python
data = [10, 20, 30, 40, 50]#no need of () and[]
s=pd.Series(data,index =[1,2,3,4,5])
print(s)
```

```python
print(s.loc[1]) #given index
print(s.iloc[1:5]) #default index
print(s.loc[1:3]) # loc doesnot index rule
```

```python
data = [10,20,30.40]
s=pd.Series(data,index = ['a','b','c'])
print(s[s<30])
print(s.loc['a':'c'])
print(s.iloc[:2])
print(s[s>30])
```

```python
import pandas as pd
sales = pd.Series([1200,850,1500,950,2000],index=['monday','tuesday','wednesday','thursday','friday'])
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```

```python
#function
import pandas as pd
data = [10, 20, 30,30, 40, 50,50]
s=pd.Series(data)
print(s)
print("with functions:")
print(s.unique()) #remove repeating numbers/data
print(s.nunique()) #return the count of unique data
print(s.value_counts())#how many times a data is repeated
```

```python
import pandas as pd
data = [10, 20, 30,None, 40, 50,50,None]
s=pd.Series(data)

print(s.isnull())
print(s.notnull())

#what is the diffrence?
```

```python
print(s.isna())
print(s.notna())
```
