```python
data = [10, 20, 30, 40, 50]#no need of () and[]
s=pd.Series(data,index =[1,2,3,4,5])
print(s)
```

```python
print(s.loc[1]) #given index
print(s.iloc[1:5]) #default index
print(s.loc[1:3]) # loc doesnot index rule
```

```python
data = [10,20,30.40]
s=pd.Series(data,index = ['a','b','c'])
print(s[s<30])
print(s.loc['a':'c'])
print(s.iloc[:2])
print(s[s>30])
```

```python
import pandas as pd
import numpy as np

data = {"name":["arya","rinza","sreya","bob","charlie"],"age":[20,21,22,23,24],"marks":[85,72,np.nan,90,65],"attendence":[95,80,75,np.nan,88],"city":["kochi","delhi","mumbai","kochi","chennai"]}
df = pd.DataFrame(data)
print(df)
```

```python
#head() and tail() by defult it gives 5 values
print("\t\t\t first 3 numbers\t\t\t")
df.head(3)

```

```python
print("\t\t\t last 3 numbers\t\t\t")
df.tail(3)
```

```python
#functions in data frame
print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
```

```python
#single cloumn
print(df.name)
print(df.marks)
print(df[['name','marks']]) #more than a single row and column
```

```python
print(df)
```

```python
#slicing
#1=iloc()
print(df.iloc[0:3]) #up to 2
#2=loc()
print('\t')
print(df.loc[1:3,['name','marks']]) #full 3 values
print("\t")
print(df.loc[1:3,['attendence','city']])
```

```python
#filtering with conditions
print("marks greater than 75:")
print(df[df['marks']>75])
print("\t")
print("\nstudents from kochi:")#\n  newline character tells the interpreter to move the cursor to the beginning of the next line
                                #This effectively creates a line break in the output, making your printed text appear on separate lines.
print(df[df['city']=='kochi'])
print("\t")
print("\nstudents with marks >70 and attendence >80")
print(df[(df['marks']>70) & (df['attendence']>80)]) #use() from multiple arguments in the same line
```

```python
#MATH OPERATIONS
print(df["marks"])
print("\t")
print("\nadd 5")
print(df['marks']+5)
print("\t")
print("\nmarks*2")
print(df['marks'] * 5)
print("\t")
print("\naverage marks")
print(df['marks'].mean)
print("\t")
print("\nmaximun marks")
print(df['marks'].max)
print("\t")
print("\nminimun marks")
print(df['marks'].min)
```

```python
#sorting
print("sort by marks")
print(df.sort_values('marks'))
print("\t.............................\t")
print(df.sort_values('marks',ascending = True))
```

```python
#rename
df_rename = df.rename(columns = {'marks':"score","attendence":"attendence_percentage"})
print("\t\trenamed dataframe\t\t")
print(df_rename)
```

```python
#check for missing value
print(df.isna())
print("\t.................\t")
print(df.isna().sum())
print("\t.................\t")
print(df.notna())
print("\t.................\t")
print("drop nan values")
print(df.dropna())
```

```python
#FORWARD\BACKWARD FILL
df_forward = df.copy()#copy dataframe
df_forward = df_forward.ffill() #syntax
print("\nFORWARD FILL:")
print(df_forward )
print("\t")
print("\t...................\t")
df_backward = df.copy()#copy dataframe
df_backward = df_backward.ffill() #syntax
print("\nBSCKWARD FILL:")
print(df_backward )
```
