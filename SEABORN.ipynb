```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = np.random.randn(1000) # Make 1000 random pretend numbers that look like they're from a bell curve. We use 'np.random.randn' because it's a quick way to get many numbers for our picture.
plt.figure(figsize = (8,6)) # Get a big piece of paper for our drawing, 8 inches wide and 6 inches tall. We use this to set how big our picture will be before we start drawing on it.
sns.histplot(data,bins = 30,kde = True,color = "skyblue",edgecolor = "darkblue")
# 'bins = 30': This means we cut our numbers into 30 different groups or buckets. It's like having 30 different boxes to sort our toys into.
# 'kde = True': This is like drawing a smooth roller coaster track over the top of our toy boxes to see the general shape of how our toys are spread out.
plt.xlabel ("values")
plt.ylabel("frequency")
plt.title("distribution of random data using seaborn")
plt.grid(axis = 'y',alpha = 0.75) # Add faint lines across the picture to help us see the height of the bars better. We set 'axis = 'y'' to only draw lines horizontally.
# 'alpha = 0.75': This makes the lines a little see-through, not too dark, so they don't hide our main drawing. It's like drawing with a light crayon instead of a super dark one.
plt.show()
```

```python
products = ["laptop","phone","tablet"]
sales = [50,80,40]
plt.pie(sales,labels = products,autopct="%4.2f%%")#write down the percentage on each slice, and make sure it has two numbers after the dot, like 4.25% or 10.00%
plt.title("sales distribution")
plt.show()
```

```python
#fill between function
#with function
print("/t/t...........\t/t")
x=(2,4,6,10)
y=(1,2,7,8)
plt.plot(x,y)
plt.fill_between(x,y)
plt.show()
#without function
print("/t/t...........\t/t")
print("without function")
x=(2,4,6,10)
y=(1,2,7,8)
plt.plot(x,y)
plt.show()
```
