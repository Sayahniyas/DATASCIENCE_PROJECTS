```python
#used to visualize data
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [2,4,6,8,10,12]
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["jan",'feb','mar','april','may']
sale = [20,30,25,40,50]
plt.plot(months,sale)

plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["jan",'feb','mar','april','may']
sale = [20,30,25,40,50]
profit = [5,8,6,12,15]
plt.plot(months,sale,label = 'sales')
plt.plot(months,profit,label = 'profit')
plt.xlabel("months")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#grids and titles
import matplotlib.pyplot as plt
months = ["jan",'feb','mar','april','may']
sale = [20,30,25,40,50]
profit = [5,8,6,12,15]
plt.plot(months,sale,label = 'sales')
plt.plot(months,profit,label = 'profit')
plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#costamizations
import matplotlib.pyplot as plt
months = ["jan",'feb','mar','april','may']
sale = [20,30,25,40,50]
profit = [5,8,6,12,15]
plt.plot(months,sale,label = 'sales',marker = 'o',color = 'red',linestyle = "--") #to customiza add them to plt.plot section here
plt.plot(months,profit,label = 'profit',marker = 'x',color = 'blue',linestyle = "-.")
plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#bar chart
import matplotlib.pyplot as plt
months = ["jan",'feb','mar','april','may']
sale = [20,30,25,40,50]
profit = [5,8,6,12,15]
plt.bar(months,sale,label = 'sales') #to customiza add them to plt.plot section here
plt.bar(months,profit,label = 'profit')
plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#scatter
import matplotlib.pyplot as plt
study_hours = [1,2,3,4,5,6]
marks = [ 40,45,55,65,75,85]
plt.scatter(study_hours,marks)
plt.title("study hours vs marks")
plt.xlabel("study hours")
plt.ylabel("marks")
plt.show()
```

```python
#histogram
import matplotlib.pyplot as plt
marks = [40,99,85,63,55,98,77,64,50,65]
plt.hist(marks)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["jan",'feb','mar','april','may']
sale = [20,30,25,40,50]
plt.pie(sale , labels = months)
plt.title("monthly sales")
plt.show()
```

```python
#add multiple graphs to one chart
import matplotlib.pyplot as plt
fig, ax = plt.subplots(2,1)
ax[0].plot([1,2,3],[10,20,30])
ax[1].bar(['A','B','C]='],[20,15,30])
plt.show()
```
